```python
# Imports
from helper_lib.data_loader import get_data_loader
from helper_lib.trainer import train_model
from helper_lib.evaluator import evaluate_model
from helper_lib.model import get_model
from helper_lib.checkpoints import load_checkpoint
from helper_lib.utils import get_device
import torch.nn as nn
import torch.optim as optim
```

```python
# Load Data
train_loader = get_data_loader('data/train', batch_size=64, train=True)
val_loader = get_data_loader('data/val', batch_size=64, train=False)
test_loader = get_data_loader('data/test', batch_size=64, train=False)
```

```python
# Init and Train Model
device = get_device()
print(f"Using device: {device}")

model = get_model("CNN")  # Choose from "FCNN", "CNN", "EnhancedCNN"
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

trained_model = train_model(
    model, train_loader, val_loader, criterion, optimizer, 
    device=device, epochs=10, checkpoint_dir='checkpoints'
)

avg_loss, accuracy = evaluate_model(trained_model, test_loader, criterion, device=device)
print(f"Test Accuracy: {accuracy:.2f}%")
```
