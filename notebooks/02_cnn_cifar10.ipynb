```python
import torch
import torch.nn as nn
import torch.optim as optim

from helper_lib.data_loader import get_data_loader
from helper_lib.evaluator import evaluate_model
from helper_lib.model import get_model
from helper_lib.trainer import train_model
from helper_lib.utils import get_device, save_model

torch.manual_seed(42)
device = get_device()
print(f"Using device: {device}")
```

```python
train_loader = get_data_loader("data", batch_size=64, train=True)
test_loader = get_data_loader("data", batch_size=64, train=False)
```

```python
model = get_model("CNN")  # one of "FCNN", "CNN", "EnhancedCNN"
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

model = train_model(model, train_loader, test_loader, criterion, optimizer,
                    device=device, epochs=10, checkpoint_dir="checkpoints")
```

```python
loss, acc = evaluate_model(model, test_loader, criterion, device=device)
print(f"Test accuracy: {acc:.2f}%")

save_model(model.cpu(), "app/models/cnn_cifar10.pth")
```
